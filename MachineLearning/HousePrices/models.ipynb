```python
import nbimporter
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.model_selection import train_test_split, cross_val_score
from evaluation import rmsle, rmsle_scorer
import numpy as np

def train_linear_regression(X_train, y_train, X_valid, y_valid):
    model = LinearRegression()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_valid)
    rmse_val = np.sqrt(np.mean((y_valid - y_pred)**2))
    rmsle_val = rmsle(y_valid, y_pred)
    return model, y_pred, rmse_val, rmsle_val

def train_gradient_boosting(X_train, y_train, X_valid=None, y_valid=None, n_estimators=800, max_depth=2, learning_rate=0.1):
    gbr = GradientBoostingRegressor(
        n_estimators=n_estimators,
        max_depth=max_depth,
        learning_rate=learning_rate,
        random_state=42
    )
    gbr.fit(X_train, y_train)
    
    if X_valid is not None:
        y_pred = gbr.predict(X_valid)
        rmse_val = np.sqrt(np.mean((y_valid - y_pred)**2))
        rmsle_val = rmsle(y_valid, y_pred)
        return gbr, y_pred, rmse_val, rmsle_val
    return gbr

def train_random_forest(X_train, y_train, X_valid=None, y_valid=None, n_estimators=200, max_depth=20):
    rfr = RandomForestRegressor(
        n_estimators=n_estimators,
        max_depth=max_depth,
        random_state=42
    )
    rfr.fit(X_train, y_train)
    
    if X_valid is not None:
        y_pred = rfr.predict(X_valid)
        rmse_val = np.sqrt(np.mean((y_valid - y_pred)**2))
        rmsle_val = rmsle(y_valid, y_pred)
        return rfr, y_pred, rmse_val, rmsle_val
    return rfr
```
