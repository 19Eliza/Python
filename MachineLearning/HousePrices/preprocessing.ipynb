```python
import nbimporter
import pandas as pd
import numpy as np

def load_data(train_path="train.csv", test_path="test.csv"):
    """Load train and test datasets."""
    train = pd.read_csv(train_path)
    test = pd.read_csv(test_path)
    
    y = train.pop('SalePrice')   # target
    id_train = train.pop('Id')   # keep train ID
    id_test = test.pop('Id')     # keep test ID
    
    return train, test, y, id_train, id_test

def handle_missing(train, test):
    """Fill missing values for categorical and numeric columns."""
    # Categorical
    categorical_cols = train.select_dtypes(include=['object']).columns
    train[categorical_cols] = train[categorical_cols].fillna(0)
    test[categorical_cols] = test[categorical_cols].fillna(0)
    
    # Numeric
    numeric_train_cols = train.select_dtypes(include=[np.number]).columns
    numeric_test_cols = test.select_dtypes(include=[np.number]).columns
    train[numeric_train_cols] = train[numeric_train_cols].fillna(0)
    test[numeric_test_cols] = test[numeric_test_cols].fillna(0)
    
    return train, test

def encode_categorical(train, test):
    """One-hot encode categorical variables."""
    train = pd.get_dummies(train, drop_first=True)
    test = pd.get_dummies(test, drop_first=True)
    
    # Align train and test to have same columns
    train, test = train.align(test, join='left', axis=1, fill_value=0)
    
    return train, test
```
