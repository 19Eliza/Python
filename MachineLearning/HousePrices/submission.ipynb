```python
import pandas as pd
import numpy as np

def create_submission(model, test_data, id_test, filename="submission.csv"):
    X_test = np.asarray(test_data)
    y_pred = model.predict(X_test)
    y_pred = np.maximum(0, y_pred)  # ensure non-negative prices

    submission = pd.DataFrame({
        "Id": id_test,
        "SalePrice": y_pred
    })
    submission.to_csv(filename, index=False)
    print(f"File {filename} has been created.")
```
