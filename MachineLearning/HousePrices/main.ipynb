```python
import nbimporter
from preprocessing import *
from models import train_linear_regression, train_gradient_boosting, train_random_forest
from evaluation import rmse, rmsle
from submission import create_submission
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

# -------------------------------
# Load and preprocess data
# -------------------------------

train_data, test_data, y, id_train, id_test = load_data()
train_data, test_data = handle_missing(train_data, test_data)
train_data, test_data = encode_categorical(train_data, test_data)

# Split for validation
X_train, X_valid, y_train, y_valid = train_test_split(train_data, y, test_size=0.2, random_state=42)

# -------------------------------
# Train and evaluate Linear Regression
# -------------------------------
lr_model, y_pred_lr, rmse_lr, rmsle_lr = train_linear_regression(X_train, y_train, X_valid, y_valid)
print(f"Linear Regression: RMSE={rmse_lr:.4f}, RMSLE={rmsle_lr:.4f}")

# -------------------------------
# Choose optimal parameters for Gradient Boosting
# -------------------------------

# -------------------------------
# Train and evaluate Gradient Boosting
# -------------------------------
gbr_model, y_pred_gbr, rmse_gbr, rmsle_gbr = train_gradient_boosting(X_train, y_train, X_valid, y_valid)
print(f"Gradient Boosting: RMSE={rmse_gbr:.4f}, RMSLE={rmsle_gbr:.4f}")

# -------------------------------
# Train and evaluate Random Forest
# -------------------------------
rfr_model, y_pred_rfr, rmse_rfr, rmsle_rfr = train_random_forest(X_train, y_train, X_valid, y_valid)
print(f"Random Forest: RMSE={rmse_rfr:.4f}, RMSLE={rmsle_rfr:.4f}")

# -------------------------------
# R² scores
# -------------------------------
print("R² scores:")
print(f"Linear Regression: {r2_score(y_valid, y_pred_lr):.4f}")
print(f"Gradient Boosting: {r2_score(y_valid, y_pred_gbr):.4f}")
print(f"Random Forest: {r2_score(y_valid, y_pred_rfr):.4f}")

# -------------------------------
# Final submission with Gradient Boosting
# -------------------------------
create_submission(gbr_model, test_data, id_test, filename="submission.csv")
```
